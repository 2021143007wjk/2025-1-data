```python
url = 'http://openapi.seoul.go.kr:8088/59527456487368693834454c474a6c/json/tbEntranceItem/1/100/'
# 보건소 데이터
response = requests.get(url)

data = response.json()
data
```

```python
temp = []
for d in data['tbEntranceItem']['row']:
  temp.append(d)
pd_data = pd.DataFrame(temp)
pd_data
```

```python
import folium
from folium.plugins import MarkerCluster

rest_map = folium.Map(location=[37.650288, 127.036860],
                      zoom_start=10,
                      zoom_control=False,
                      control_scale=True)

mc = MarkerCluster().add_to(rest_map)

for n, row in pd_data.iterrows():
  folium.Marker(location=[row['LAT'], row['LOT']],
                popup=folium.Popup(row['FCLT_NM'], max_width=200)).add_to(mc)

rest_map
```
